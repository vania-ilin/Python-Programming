**Task 2.** Analyze the data for gaps. Determine if there are, and if so, how many. Apply different approaches to eliminate gaps:

* Deleting records with gaps
* Filling gaps with different methods (constant - average, median, mode; averaging neighboring values)

```python
#197-187 = 12 skips

# method isna() - returns a bitmap of gaps. NA (not-a-number)
print(sales.isna())
```

```python
# you can sum the true(1) values in the resulting bitmap. Total = 12 passes
print(sales.isna().sum())
```

```python
# selection of values by true values. It turns out a series of only passes
sales_with_na = sales[sales.isna()]
sales_with_na
```

```python
# another way is to algorithmically iterate over the elements and check if there is a gap
na_count = 0 # skip counter variable
for value in sales:  # iterate over values
  if pd.isna(value): # if the current element is a blank
    na_count = na_count + 1
print(na_count) # as you can see, the counted number of passes coincided with the previous ones
```

```python
# a) delete records with gaps
sales_without_na = sales.dropna(inplace=False)  # "don't save" the changes to data (sales), and construct a new object. If you want to save changes then inplace=TRUE
sales_without_na  # now length = 187
```

```python
# b) filling in the gaps with "mean = 9.8". replacement method .replace().
# You can also replace gaps with "median". Maybe for fashion
sales.replace(np.nan, sales.mean(), inplace=False) # inplace default = false
```

```python
# c) filling in the gaps using the "interpolation" method (averaging adjacent values - replacing the average of adjacent values \u200b\u200b(previous and subsequent))
sales.interpolate() # default "simple linear interpolation" sales.interpolate(method="linear", inplace=False)

# DETAILED DISCUSSION HOW IT WORKS:
# how to make sure that you really averaged the neighbors? Select values with gaps
# and look at previous and next

# save indexes with gaps separately
na_index = sales[sales.isna()].index
print(na_index)

# look at previous and next gaps
# loop through them
for index in na_index:
  # display the previous element and the skip itself
  print(sales[index - 1], sales[index], end=" ")
# and the next one will be displayed by condition: only if there is such an index at all
   # (otherwise we will run into an error: the last gap is also the last element
   # series, no next, so no element at index + 1)
  if index + 1 in sales.index:
    print(sales[index + 1])
print("\n")

# now interpolate, save the result inplace=True
sales.interpolate(method="linear", inplace=True)

# and look again - did you really average it? For example 1st pass: (31.37 + 30.01) / 2 = 30.69

for index in na_index:
  print(sales[index - 1], sales[index], end=" ")
  if index + 1 in sales.index:
    print(sales[index + 1])

# as you can see, the gaps were indeed replaced by the formula
# (prev. element + next element) / 2
```

**Task 3.** Filling in the gaps of categorical (non-numerical) features.

Let's save the "Publisher" column in a separate variable. Check it for gaps. If there are gaps, restore them in different ways: (a) fashion; (b) filling with a value with a probability from the "generator of values from the normal distribution law" (method np.random.choice())

```python
publ = df.Publisher
publ  # as you can see, there are gaps - in particular, the last element is also missing here
publ.isna().sum()  # is the only pass
```

```python
# a) restore the omission by the mode (the qualitative (categorical), i.e., non-numeric features also have a mode)
# mode - the most frequently occurring value(s) in the sample
publ.replace(np.nan, publ.mode()[0]) # publ.mode()[0] = Nintendo
```

```python
# b) filling with a value from the "generator of values ​​from the normal distribution law"
# i.e. we will choose a value from the list resp. their probabilities
# How to determine what Nintendo is most common?
publ.value_counts() # method will return "frequency table of absolute frequencies" i.e. number of values in the sample
publ.value_counts(normalize = True) # will produce a "frequency table of relative frequencies" i.e. proportion of the value in the sample
# i.e. obtained a ready distribution law of categorical value.
# For example, we see that Nintendo is the most common in the sample (share, or relative frequency, = 0.419).
```

```python
# save the "frequency table of relative frequencies" data into a separate variable
freq_table = publ.value_counts(normalize=True) 
# we see that this is a series with indexes - the names of publishers and values - their frequencies

# Let's use the np.random.choice() function from the numpy library:
# it selects (generates) a value from the list with the probability specified in the column publ.value_counts(normalize = True)
# but the values of publishers are in our indices, and the probabilities themselves are values
np.random.choice(freq_table.index, p=freq_table) # 1 parameter - values of categorical features; 2nd parameter - probability (from English probability)
# the Nintendo value will appear in 41% of the calls to the choice function
```

```python
# now replace the gap
publ.replace(np.nan, np.random.choice(freq_table.index, p=freq_table), inplace=True) # .replace() makes changes to publ because inplace=True. New object is not constructed
df.info()  # as you can see, there are no gaps in the Publisher column (there are 199 lines, but there were 198 lines). Moreover, we saved the values in the df table, because inplace=True
# !!!!!!!despite the fact that they were accessing not through the dataframe object, but directly to the column-series objects.

# indentation: the assignment operation for "data structures" in Python does not work like a copy, but like a rearrangement of a reference.
# therefore, when copying a separate column into a new variable, we simply refer from the new variable to this dataframe column
# To create a physical copy, you need the .copy() method
```

# **Outliers in statistics**

**Task 4.** Analyze global sales data for outliers. Apply several methods for detecting outliers:

* N-sigma rule

3-sigma rule: 99.7% of all values of a normally distributed random variable deviate from its mathematical expectation by no more than 3 standard deviations. The remaining 0.3% is emissions


* Interquartile range

```python
# a) Identification of outliers. "N-sigma rule". see "normal distribution density function plot"

# To apply the N-sigma rule, you must first calculate:
# 1) asymmetry: - for symmetric N=3 (// for symmetric distributions, the arithmetic mean, mode and median are equal),
# - for asymmetric distribution N=5
# 2) range limits mean ± N sigma (sigma is standard deviation. i.e. .std() method)

# How to determine if the distribution is symmetrical or not? We are looking for the skewness coefficient - the skew () method. If modulo up to 0.5,
# we will assume that the distribution is symmetric and N = 3
if abs(sales.skew()) <= 0.5: # ideally d.b. coefficient=0
  N = 3
  print("The distribution is symmetrical, N = 3")
else:
  N = 5
  print("The distribution is not symmetrical, N = 5")

# check for anomaly (outlier) maximum i.e. extreme right point
sigma = sales.sort_values().iloc[:-1].std()  # check for anomaly (outlier) maximum i.e. extreme right point
mean = sales.sort_values().iloc[:-1].mean()
# iloc - access by serial numbers starting from zero; loc - referencing by indexes
right = mean + N * sigma
if sales.max() > right:
  print("value", sales.max(), "is an outlier according to the criterion " + str(N) + "-sigma")  # cast the number to the string with the str() function. Or f-string can be used

# check for anomaly (outlier) at least i.e. extreme left point
sigma = sales.sort_values().iloc[1:].std()  # similarly remove the minimum itself from the sample
mean = sales.sort_values().iloc[1:].mean()
left = mean - N * sigma
if sales.min() < left:
  print("value", sales.min(), "is an outlier according to the criterion " + str(N) + "-сигма")

sales.sort_values() # print for clarity
# Interpretation of the results: the maximum value of 82.74 here is an outlier (by the way, the previous one before the maximum can also be checked - suddenly also an outlier),
# minimal - turned out not to be an outlier (further on, the previous ones can no longer be checked)
```

```python
# B) Identification of outliers. Rule "Interquartile range"

# Fact: 50% of the sample items belong to the interval [Q25, Q75]
# Rule: elements that do not fall into the interval [Q25 - 1.5*(Q75-Q25); Q75 + 1.5*(Q75-Q25)] can be declared outliers

q1 = sales.quantile(0.25)  # первый квартиль т.е. уровня 25 (= 5.48)
q3 = sales.quantile(0.75)  # third quartile i.e. level 75 (= 10.73)
iqr = q3 - q1  # actually IQR (InterQuartile Range) from English. "interquartile range" (= 5.25)
low = q1 - 1.5 * iqr  # (lower limit of acceptable range, outliers) (= -2.39)
high = q3 + 1.5 * iqr  # (upper range limit) (= 18.60)

# there are a lot of outliers, but we will check for anomalies only the maximum
if sales.max() > high:
  print("value", sales.max(), "is an outlier according to the interquartile range criterion")

# и проверим на аномальность минимум
if sales.min() < low:
  print("value", sales.min(), "is an outlier according to the interquartile range criterion")
```

**Task 5.** Eliminate outliers:

* (a) by the same methods as omissions
* (b) smoothing (for example, changing to 1st for the minimum and 99th for the maximum percentile)

```python
sales

for value in sales:
  print (value, "is an outlier", not low <= value <= high)
# we see that from a value of 20.22 and above - outliers
```

```python
# first get emission indices
o_index = []
for index in sales.index:
  # if out of range
  if not (low <= sales[index] <= high):
    o_index.append(index)  # in general, indexes can be duplicated, but not in our data

sales.head(25)  # see the entire selection: outlier elements that belong [-2.39 to 18.60]
```

```python
# Method(a). Idea: replacing outliers with gaps and applying gap recovery methods
# sales[o_index] = np.nan
# and then, for example, like this
# sales.interpolate()
# but we won't..
```

```python
# Method (b). Smoothing. Replace outliers with quantiles
perc99 = sales.quantile(0.99)  # quantile 0.99 (99th percentile) i.e. this is such a value that 99% of the sample is located to the left (smaller) of it. And 1% above it. Those. also large, but less than the maximum
print("99th percentile = ",  perc99)
perc1 = sales.quantile(0.01)  # quantile 0.01 (1st percentile)
print("1st percentile = ",  perc1)
for index in o_index:  # iterate over outlier indices only
  if sales[index] > high:  # if the outlier is greater than the upper bound
    sales.replace(sales[index], perc99, inplace=True)  # replace with 99th percentile
  else:  # otherwise the outlier is less than the lower bound - we only iterate over the outliers
    sales.replace(sales[index], perc1, inplace=True) # replace with 1st percentile

sales
```
